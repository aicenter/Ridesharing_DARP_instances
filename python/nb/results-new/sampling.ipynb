## setup

### prepare occupancy dataframe

```python
area = 'Manhattan'
```

```python
sam_df = load_occupancies_in_dir(RESULTS_PATH / area)
sam_df['area'] = area
```

```python
sam_df['cost_per_request'] = sam_df['cost_minutes'] * 60 / sam_df['req_count']
sam_df['area_short'] = sam_df['area'].map({
    'Manhattan': 'MH'
})
area_order = {
    'Manhattan': 3,
}
sam_df['area_order'] = sam_df['area'].map(area_order)
sam_df.sort_values(by=['area_order', 'duration_minutes', 'max_delay', 'method'], inplace=True)
sam_df['method'] = sam_df['method'].replace('halns', 'alns')
```

# Cost

## Capacity

```python
chosen_df = sam_df[sam_df['method'] == 'ih'].sort_values(by=['capacity', 'duration_minutes', 'max_delay', 'sample'])
```

```python
fig = px.line(
    chosen_df,
    x='sample',
    y='cost_per_request',
    color='capacity',
    facet_row='max_delay',
    facet_col='duration_minutes',
    markers=True,
)

# Titles
fig.for_each_yaxis(lambda y: y.update(title=''))  # Remove titles
fig.for_each_xaxis(lambda x: x.update(title=''))  # Remove titles
fig.add_annotation(x=.45, y=-0.15, text="Demand sampling factor", xref="paper", yref="paper", showarrow=False)  # x axis
fig.add_annotation(x=-0.08, y=0.5, text="Average cost per request [s]", textangle=-90, xref="paper", yref="paper", showarrow=False)  # y axis
fig.add_annotation(x=0.45, y=1.12, text="Instance length [min]", xref="paper", yref="paper", showarrow=False)  # Facet col
fig.add_annotation(x=1.05, y=0.5, text="Maximum delay [s]", xref="paper", yref="paper", showarrow=False, textangle=90)  # Facet row
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))

fig.update_layout(legend=dict(orientation="h", yanchor="bottom", y=1.05, xanchor="right", x=1.0), font=dict(size=28), height=900, width=1500,
)
fig.for_each_xaxis(lambda x: x.update(tickmode='linear', tick0=0, dtick=0.25, tickfont=dict(size=18), tickformat=".0%"))

# fig.write_image(f"{IMG_PATH}/avg_cost_sample_capacity.pdf")
fig.show()
```

# Average trip duration - box

```python
cost_df = sam_df[['sample', 'max_delay', 'avg_trip_duration', 'duration_minutes', 'method', 'capacity']].drop_duplicates().sort_values(by=['sample', 'duration_minutes', 'capacity'])
N = cost_df['sample'].nunique()
color_seq = ['hsl(' + str(h) + ',60%,55%)' for h in np.linspace(15, 345, N)]
```

```python
fig = px.box(
    cost_df,
    y='avg_trip_duration',
    x='sample',
    color='sample',
    facet_col='max_delay',
    facet_row='capacity',
    notched=True,
    color_discrete_sequence=color_seq
    )

# small figures
fig.for_each_yaxis(lambda y: y.update(title=''))  # Remove titles
fig.for_each_xaxis(lambda x: x.update(title=''))  # Remove titles

# titles
fig.add_annotation(x=.49, y=1.065, text="Maximum delay [s]", xref="paper", yref="paper", showarrow=False) # facet col title
fig.add_annotation(x=0.5, y=-0.085, text="Average trip duration [s]", xref="paper", yref="paper", showarrow=False) # x axis title
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))
fig.update_layout(legend_title_text='Sampling factor', legend=dict(orientation="h", yanchor="bottom", y=-.15, xanchor="right", x=1.0), font=dict(size=25), height=1100, width=1600)

fig.for_each_xaxis(lambda x: x.update(tickmode='linear', tick0=0, dtick=0.25, tickfont=dict(size=18), tickformat=".0%"))
fig.for_each_trace(lambda t: t.update(name=f"{float(t.name) * 100:.0f}%"))

# fig.write_image(f"{IMG_PATH}/avg_duration_box_sample.pdf")
fig.show()
```

# Occupancy

```python
selected_samples = [0.2, 0.4, 0.6, 0.8, 1.25, 1.5, 1.75, 2.0] # filter some for simplicity
filtered_df = sam_df[sam_df['sample'].isin(selected_samples)].copy()
filtered_df.loc[:, 'occupancy_grouped'] = filtered_df['occupancy'].apply(lambda x: '6+' if x >= 6 else str(x))
```

```python
cap_df = filtered_df[['sample', 'vehicle_hours', 'occupancy_grouped', 'method', 'capacity']].drop_duplicates()
cap_df = cap_df.groupby(['sample', 'method', 'occupancy_grouped', 'capacity'])['vehicle_hours'].sum().reset_index()
total_vehicle_hours = cap_df.groupby(['sample', 'method', 'capacity'])['vehicle_hours'].transform('sum')
cap_df['vehicle_hours_percentage'] = (cap_df['vehicle_hours'] / total_vehicle_hours) * 100
cap_df = cap_df.sort_values(by=['occupancy_grouped', 'capacity'])
```

```python
fig = px.histogram(
    cap_df,
    x='occupancy_grouped',
    y='vehicle_hours_percentage',
    color='method',
    barmode='group',
    facet_col='sample',
    facet_row='capacity',
)

# small figures
fig.for_each_yaxis(lambda y: y.update(title=''))  # Remove titles
fig.for_each_xaxis(lambda x: x.update(title=''))  # Remove titles

# titles
fig.add_annotation(x=0.4, y=1.1, text="Sample size", xref="paper", yref="paper", showarrow=False) # facet column title
fig.add_annotation(x=1.04, y=0.5, text="Max capacity", xref="paper", yref="paper", showarrow=False, textangle=90) # facet row title
fig.add_annotation(x=-0.06, y=0.5, text="Vehicle hours [%]", textangle=-90, xref="paper", yref="paper", showarrow=False) # y axis title
fig.add_annotation(x=0.5, y=-0.07, text="Occupancy", xref="paper", yref="paper", showarrow=False) # x axis title
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))
fig.update_layout(legend=dict(orientation="h", yanchor="bottom", y=1.05, xanchor="right", x=1.03), font=dict(size=28), height=1000, width=1400,)

fig.update_xaxes(tickmode='linear', tick0=0, dtick=1, tickfont=dict(size=18), showticklabels=True)

# fig.write_image(f"{IMG_PATH}/hours_occupancy_sample.pdf")
fig.show()
```
